# BƯỚC 1: Nạp dữ liệu và kiểm tra ban đầu (Data Loading & Initial Exploration)

```python
import numpy as np
import pandas as pd
from pathlib import Path

# Cấu hình hiển thị pandas
pd.set_option("display.max_columns", 100)
pd.set_option("display.max_rows", 100)
pd.set_option("display.float_format", "{:,.3f}".format)
```

```python
# Đường dẫn tới file dữ liệu gốc
DATA_PATH = Path("../data/raw/accepted_2007_to_2018Q4.csv")
if not DATA_PATH.exists():
    DATA_PATH = Path("data/raw/accepted_2007_to_2018Q4.csv")

df_raw = pd.read_csv(DATA_PATH, low_memory=False)
```

```python
def summarize_columns(df):
    first_valid = df.apply(lambda s: s.dropna().iloc[0] if s.notna().any() else None)
    summary = pd.DataFrame({
        "dtype": df.dtypes.astype(str),
        "missing_count": df.isna().sum(),
        "missing_pct": (df.isna().mean() * 100),
        "nunique": df.nunique(),
        "sample_val": first_valid
    })
    return summary
```

## 2 Feature selection

### Feature type handling

```python
# Filter the DataFrame label to only include 'Fully Paid' and 'Charged Off' loans
df = df_raw.loc[(df_raw['loan_status'].isin(['Fully Paid', 'Charged Off']))]

df.shape
```

```python
# Chuyển đổi issue_d sang datetime
df["issue_date"] = pd.to_datetime(df["issue_d"], format="%b-%Y")

# In khoảng thời gian dữ liệu bao phủ
min_date = df["issue_date"].min().strftime("%m/%Y")
max_date = df["issue_date"].max().strftime("%m/%Y")
print(f"Dữ liệu trải dài từ tháng: {min_date} đến tháng: {max_date}")
```

```python
pd.set_option("display.float_format", "{:,.2f}".format)

df["target"] = df["loan_status"].map({"Fully Paid": 0.0, "Charged Off": 1.0})
df["issue_year"] = df["issue_date"].dt.year
df["term_months"] = df["term"].str.extract(r"(\d+)")[0].astype(float)

yearly_stats = df.groupby("issue_year").agg(
    total_loans=("target", "count"),
    default_rate=("target", lambda x: x.mean() * 100),
    term_60m_pct=("term_months", lambda x: (x == 60).mean() * 100)
)

term_default = (
    df.groupby(["issue_year", "term_months"])["target"]
    .mean()
    .unstack("term_months") * 100
).rename(columns={36.0: "default_rate_36m", 60.0: "default_rate_60m"})

yearly_stats = yearly_stats.join(term_default)[
    ["total_loans", "default_rate", "default_rate_36m", "default_rate_60m", "term_60m_pct"]
].round(2)
yearly_stats.index = yearly_stats.index.astype(int)
print("Thống kê khoản vay theo từng năm phát hành và kỳ hạn:")
yearly_stats
```

```python
# Chỉ giữ các khoản vay đã đáo hạn (tháng giải ngân + kỳ hạn <= tháng cuối của dữ liệu)
snapshot = df["issue_date"].max()
snapshot_month = snapshot.year * 12 + snapshot.month
end_month = (df["issue_date"].dt.year * 12 + df["issue_date"].dt.month) + df["term_months"]

df = df[end_month <= snapshot_month].copy()

# Xóa 3 cột cũ đã xử lý (loan_status, issue_d, term) và cột tạm (issue_year)
df = df.drop(columns=["loan_status", "issue_d", "term", "issue_year"])

print(f"Số khoản vay còn lại sau khi lọc đã đáo hạn: {len(df):,} dòng")
df.shape
```

```python
# Remove columns with more than 80% missing values
heavy_missing = col_summary[col_summary["missing_pct"] > 80].index.tolist()
print(f"Loại bỏ {len(heavy_missing)} cột có tỷ lệ missing > 80%")

df = df.drop(columns=heavy_missing)
df.shape
```

```python
# Remove duplicate features
duplicate_features = df.columns[df.columns.duplicated()].tolist()
print(f"Loại bỏ {len(duplicate_features)} cột trùng lặp")
df = df.drop(columns=duplicate_features)
df.shape
```

```python
# Remove duplicate rows
df = df.drop_duplicates()
df.shape    
```

### Feature filtering

```python
# 1. Loại bỏ các cột hằng số (chỉ có <= 1 giá trị duy nhất)
constant_features = [col for col in df.columns if df[col].nunique(dropna=True) <= 1]
print(f"Loại bỏ {len(constant_features)} cột hằng số: {constant_features}")

df = df.drop(columns=constant_features)
df.shape
```

```python
# 2. Loại bỏ cột định danh (url) và các cột text tự do có quá nhiều giá trị rời rạc
high_card_cols = [col for col in ["url", "emp_title", "title", "zip_code"] if col in df.columns]
print(f"Loại bỏ {len(high_card_cols)} cột định danh / text nhiều giá trị: {high_card_cols}")

df = df.drop(columns=high_card_cols)
df.shape
```

```python
# 3. Loại bỏ các cột gây rò rỉ dữ liệu (Data Leakage - chỉ phát sinh sau khi đã giải ngân)
LEAKAGE_COLS = [
    "funded_amnt", "funded_amnt_inv", "out_prncp", "out_prncp_inv",
    "total_pymnt", "total_pymnt_inv", "total_rec_prncp", "total_rec_int",
    "total_rec_late_fee", "recoveries", "collection_recovery_fee",
    "last_pymnt_d", "last_pymnt_amnt", "last_credit_pull_d",
    "last_fico_range_high", "last_fico_range_low", "debt_settlement_flag"
]
cols_leakage = [col for col in LEAKAGE_COLS if col in df.columns]
print(f"Loại bỏ {len(cols_leakage)} cột rò rỉ sau giải ngân (leakage): {cols_leakage}")

df = df.drop(columns=cols_leakage)
df.shape
```

```python
# 4. Xử lý các cột dạng chữ nhưng mang ý nghĩa số / thời gian (emp_length, earliest_cr_line)
# - emp_length: "< 1 year" -> 0, "1 year".."9 years" -> 1..9, "10+ years" -> 10
emp_clean = df["emp_length"].replace({"< 1 year": "0 years"})
df["emp_length_years"] = emp_clean.str.extract(r"(\d+)")[0].astype(float)

# - earliest_cr_line: đổi sang số năm lịch sử tín dụng tính đến ngày vay (issue_date)
earliest_date = pd.to_datetime(df["earliest_cr_line"], format="%b-%Y")
df["credit_history_years"] = ((df["issue_date"] - earliest_date).dt.days / 365.25).round(2)

# Xóa 2 cột chữ cũ sau khi đã chuyển sang dạng số
df = df.drop(columns=["emp_length", "earliest_cr_line"])

print("Các cột dạng chữ (str/object) còn lại trong df:")
print(df.select_dtypes(include=["object", "string"]).columns.tolist())
df[["emp_length_years", "credit_history_years"]].head()
```

# BƯỚC 2: EDA thời gian & Chia tập Train / Val / Test (70/15/15) theo Time-Series

```python
# Sắp xếp theo thời gian tăng dần
df_sorted = df.sort_values(by=["issue_date", "id"]).reset_index(drop=True)

# Xác định vị trí cắt dòng 70% / 15% / 15%
idx_train_end = int(len(df_sorted) * 0.70)
idx_val_end = int(len(df_sorted) * 0.85)

print(f"Chỉ số cắt tập Train: từ dòng 0 đến {idx_train_end:,}")
print(f"Chỉ số cắt tập Val:   từ dòng {idx_train_end:,} đến {idx_val_end:,}")
print(f"Chỉ số cắt tập Test:  từ dòng {idx_val_end:,} đến {len(df_sorted):,}")
```

# BƯỚC 3: Sàng lọc 29 Feature hợp lệ tại thời điểm nộp đơn (Application-Time Whitelist)

```python
# Whitelist 29 feature biết được tại thời điểm nộp đơn (Application Time)
FEATURES_29 = [
    # Nhóm 1: Thông tin khoản vay đề nghị (9 cột)
    "loan_amnt", "term_months", "int_rate", "installment", "grade", "sub_grade", 
    "purpose", "application_type", "initial_list_status",
    
    # Nhóm 2: Hồ sơ người vay tự khai (6 cột)
    "emp_length_years", "home_ownership", "annual_inc", "verification_status", 
    "addr_state", "dti",
    
    # Nhóm 3: Lịch sử tín dụng tra cứu từ CIC/Bureau lúc nộp đơn (14 cột)
    "delinq_2yrs", "credit_history_years", "fico_range_low", "fico_range_high", 
    "inq_last_6mths", "open_acc", "pub_rec", "revol_bal", "revol_util", 
    "total_acc", "mort_acc", "pub_rec_bankruptcies", "tax_liens", "mths_since_last_delinq"
]

TARGET_COL = "target"
ALL_30_COLS = FEATURES_29 + [TARGET_COL]

print(f"Tổng số feature lựa chọn: {len(FEATURES_29)} feature")
print(f"Tổng số cột đưa vào xử lý: {len(ALL_30_COLS)} cột (gồm cả biến target)")
```

```python
# Chỉ giữ lại 30 cột hợp lệ
train_30 = train_df[ALL_30_COLS].copy()
val_30 = val_df[ALL_30_COLS].copy()
test_30 = test_df[ALL_30_COLS].copy()

# Giữ thêm cột issue_date cho train/val/test để tính số năm lịch sử tín dụng ở bước sau
train_30["issue_date"] = train_df["issue_date"]
val_30["issue_date"] = val_df["issue_date"]
test_30["issue_date"] = test_df["issue_date"]
```

```python
print("Kích thước tập Train:", train_30.shape)
print("Kích thước tập Val:  ", val_30.shape)
print("Kích thước tập Test: ", test_30.shape)

assert all(col in train_30.columns for col in ALL_30_COLS), "Lỗi: Thiếu cột trong danh sách 30 cột!"
print(" Xác nhận: 3 tập Train/Val/Test đã được lọc chuẩn theo danh sách 29 feature + 1 target!")
```

```python
train_missing_29 = train_30[FEATURES_29].isna().mean() * 100
missing_report = train_missing_29[train_missing_29 > 0].sort_values(ascending=False).to_frame("missing_%")

print(f"Số feature có giá trị thiếu trên tập Train: {len(missing_report)} / 29 feature")
print(missing_report)
```

# BƯỚC 4: Tiền xử lý & Xây dựng Scikit-Learn Pipeline chống Data Leakage

```python
def process_features(df):
    df = df.copy()
    
    # 1. Trích xuất số tháng vay từ term nếu chưa xử lý
    if "term" in df.columns:
        df["term_months"] = df["term"].astype(str).str.extract(r"(\d+)")[0].astype(float)
    
    # 2. Trích xuất số năm công tác từ emp_length nếu chưa xử lý
    if "emp_length" in df.columns:
        emp_clean = df["emp_length"].replace({"< 1 year": "0 years"})
        df["emp_length_years"] = emp_clean.astype(str).str.extract(r"(\d+)")[0].astype(float)
    
    # 3. Tính độ dài lịch sử tín dụng (tính theo năm) nếu chưa xử lý
    if "earliest_cr_line" in df.columns:
        earliest_date = pd.to_datetime(df["earliest_cr_line"], format="%b-%Y")
        df["credit_history_years"] = (df["issue_date"] - earliest_date).dt.days / 365.25
    
    # 4. Gộp FICO cận dưới và cận trên thành điểm trung bình
    df["fico_avg"] = (df["fico_range_low"] + df["fico_range_high"]) / 2
    
    # 5. Tỷ lệ số tiền vay trên thu nhập hàng năm (Leverage ratio)
    df["loan_to_income"] = df["loan_amnt"] / (df["annual_inc"] + 1)
    
    return df
```

```python
train_processed = process_features(train_30)
val_processed = process_features(val_30)
test_processed = process_features(test_30)

print("Đã xử lý xong các đặc trưng số và text cho cả 3 tập!")
```

```python
# Danh sách biến số (đã thay thế các cột raw bằng các cột đã parse)
NUM_COLS = [
    "loan_amnt", "int_rate", "installment", "annual_inc", "dti",
    "delinq_2yrs", "inq_last_6mths", "open_acc", "pub_rec", "revol_bal",
    "revol_util", "total_acc", "mort_acc", "pub_rec_bankruptcies", "tax_liens",
    "mths_since_last_delinq", "term_months", "emp_length_years",
    "credit_history_years", "fico_avg", "loan_to_income"
]

# Danh sách biến phân loại (Categorical)
CAT_COLS = [
    "grade", "sub_grade", "purpose", "application_type",
    "initial_list_status", "home_ownership", "verification_status", "addr_state"
]

print(f"Số lượng biến số: {len(NUM_COLS)}")
print(f"Số lượng biến phân loại: {len(CAT_COLS)}")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

numeric_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="constant", fill_value="missing")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])

preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_pipe, NUM_COLS),
    ("cat", categorical_pipe, CAT_COLS)
])
```

```python
FEATURE_INPUTS = NUM_COLS + CAT_COLS

# Tách X và y
X_train_raw = train_processed[FEATURE_INPUTS]
y_train = train_processed["target"].values

X_val_raw = val_processed[FEATURE_INPUTS]
y_val = val_processed["target"].values

X_test_raw = test_processed[FEATURE_INPUTS]
y_test = test_processed["target"].values

# Fit duy nhất trên Train
preprocessor.fit(X_train_raw)

# Transform sang 3 tập
X_train = preprocessor.transform(X_train_raw)
X_val = preprocessor.transform(X_val_raw)
X_test = preprocessor.transform(X_test_raw)

print(f"Kích thước ma trận Train sau mã hóa: {X_train.shape}")
print(f"Kích thước ma trận Val sau mã hóa:   {X_val.shape}")
print(f"Kích thước ma trận Test sau mã hóa:  {X_test.shape}")
```

```python
assert not np.isnan(X_train).any(), "Lỗi: Còn NaN trong Train!"
assert not np.isnan(X_val).any(), "Lỗi: Còn NaN trong Val!"
assert not np.isnan(X_test).any(), "Lỗi: Còn NaN trong Test!"

print(" Xác nhận kiểm tra: Toàn bộ ma trận đã sạch sẽ 100%, sẵn sàng đưa vào mô hình!")
```

# BƯỚC 5: Đánh giá Baseline B1 & Huấn luyện Mô hình Logistic Regression

```python
from sklearn.metrics import roc_auc_score

# Baseline B1: xếp hạng rủi ro của sub_grade (A1 -> G5)
sub_grade_order = sorted(train_processed["sub_grade"].dropna().unique())
sub_grade_map = {grade: i for i, grade in enumerate(sub_grade_order)}

val_subgrade_score = val_processed["sub_grade"].map(sub_grade_map)
auc_baseline = roc_auc_score(y_val, val_subgrade_score)

print(f" ROC-AUC của Baseline B1 (Lending Club sub_grade trên tập Val): {auc_baseline:.4f}")
```

```python
from sklearn.linear_model import LogisticRegression

# Huấn luyện Logistic Regression
lr_model = LogisticRegression(max_iter=1000, random_state=42)
lr_model.fit(X_train, y_train)

print(" Đã hoàn thành huấn luyện Logistic Regression!")
```

```python
# Dự báo xác suất vỡ nợ trên tập Val
val_preds = lr_model.predict_proba(X_val)[:, 1]

auc_lr_val = roc_auc_score(y_val, val_preds)
lift_val = auc_lr_val - auc_baseline

print(f" ROC-AUC của Logistic Regression trên tập Val: {auc_lr_val:.4f}")
print(f" Mức cải thiện (Lift) so với Baseline sub_grade: +{lift_val:.4f}")

if lift_val > 0:
    print(" Kết luận: Mô hình của chúng ta thực sự mang lại giá trị cao hơn quy tắc có sẵn của Lending Club!")
```

```python
# Đánh giá kiểm chứng độc lập cuối cùng trên tập Test (Out-of-Time Test)
test_subgrade_score = test_processed["sub_grade"].map(sub_grade_map)
auc_baseline_test = roc_auc_score(y_test, test_subgrade_score)

test_preds = lr_model.predict_proba(X_test)[:, 1]
auc_lr_test = roc_auc_score(y_test, test_preds)
lift_test = auc_lr_test - auc_baseline_test

print(f" ROC-AUC của Baseline B1 trên tập Test:            {auc_baseline_test:.4f}")
print(f" ROC-AUC của Logistic Regression trên tập Test:     {auc_lr_test:.4f}")
print(f" Mức cải thiện (Lift) trên tập Test:               +{lift_test:.4f}")
```
