# 1. Tổng Quan Bài Toán & Từ Điển Dữ Liệu (Problem Overview & Data Dictionary)

## 1.1 Mục tiêu nghiệp vụ và phát biểu bài toán rủi ro tín dụng

- **Đầu vào:** Một khoản vay cá nhân vừa được nộp trên nền tảng Lending Club. Lúc này ta **chỉ biết** thông tin người vay tự khai và thông tin do trung tâm thông tin tín dụng (credit bureau) trả về tại thời điểm xét duyệt.
- **Đầu ra:** Điểm xác suất $P(\text{vỡ nợ})$ để xếp hạng rủi ro hồ sơ.
- **Định nghĩa vỡ nợ (nhãn):** Khoản vay bị coi là vỡ nợ nếu rơi vào trạng thái `Charged Off` (xóa nợ / mất vốn). Không tính các khoản đang vay dở dang (Current, In Grace Period, Late) vì chưa biết kết cục cuối cùng.
- **Yêu cầu quan trọng nhất:** Không được dùng bất kỳ thông tin nào phát sinh **sau** thời điểm duyệt vay (chống rò rỉ dữ liệu - data leakage).

Hàm dưới đây ghép tám nhóm thành một bảng từ điển duy nhất, mỗi cột của file gốc một dòng.

```python
def build_dictionary(groups):
    rows = []
    for group_name, cols in groups:
        for col, (meaning, known) in cols.items():
            rows.append({
                "column": col,
                "group": group_name,
                "meaning": meaning,
                "known_at_application": known,
            })
    return pd.DataFrame(rows).set_index("column")
```

## 3.3 Quy tắc các khoản vay đã đáo hạn (Matured Loans Rule)

```python
SNAPSHOT = df_raw["issue_date"].max()
print("Latest month in data (SNAPSHOT):", SNAPSHOT.strftime("%Y-%m"))
```

```python

def month_number(dates):
    return dates.dt.year * 12 + dates.dt.month
```

```python
def is_matured(issue_date, term_months, snapshot):
    end_month = month_number(issue_date) + term_months
    return end_month <= snapshot.year * 12 + snapshot.month
```

## 3.5 Phân chia tập Train / Val / Test theo Out-of-Time (70 / 15 / 15)

```python
def time_split(df, train_share=0.70, val_share=0.15):
    df = df.sort_values(["issue_date", "id"]).reset_index(drop=True)
    cut_val = df["issue_date"].iloc[int(len(df) * train_share)]
    cut_test = df["issue_date"].iloc[int(len(df) * (train_share + val_share))]
    train = df[df["issue_date"] < cut_val]
    val = df[(df["issue_date"] >= cut_val) & (df["issue_date"] < cut_test)]
    test = df[df["issue_date"] >= cut_test]
    return train, val, test, cut_val, cut_test
```

```python
train, val, test, CUT_VAL, CUT_TEST = time_split(loans, 0.70, 0.15)
print("Ngày bắt đầu tập validation:", CUT_VAL.date())
print("Ngày bắt đầu tập test:", CUT_TEST.date())
```
